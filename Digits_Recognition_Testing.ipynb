```python
import cv2
import matplotlib.pyplot as plt
```

```python
modified_image = cv2.imread("modified.PNG",cv2.IMREAD_GRAYSCALE)
```

```python
modified_image
```

```python
img_resize = cv2.bitwise_not(modified_image)
```

```python
img_resize
```

```python
plt.matshow(img_resize,cmap="gray")
```

```python
img_resize = cv2.resize(img_resize,(28,28),interpolation = cv2.INTER_LINEAR)
img_resize.shape
```

```python
img_resize_flat = img_resize.reshape(1,28*28)
img_resize_flat.shape
```

```python
import joblib as jbl 
model = jbl.load("Ashik_job")
```

```python
model.predict(img_resize_flat)
```
